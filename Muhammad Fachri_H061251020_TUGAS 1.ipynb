### Penjelasan Elemen Kerangka R–T–C–I–K–O–Q–V

* **R (Role):** Posisi atau sudut pandang keahlian yang diemban oleh kecerdasan buatan.
* **T (Task):** Sasaran pekerjaan utama yang wajib diselesaikan sistem.
* **C (Context):** Latar belakang permasalahan serta skenario di balik studi kasus.
* **I (Input):** Kumpulan parameter, data, atau variabel awal yang dilampirkan.
* **K (Constraints):** Ketentuan pembatas, larangan, serta aturan main operasional.
* **O (Output):** Bentuk susunan dan format laporan hasil akhir yang diinginkan.
* **Q (Quality Criteria):** Standar mutu dan tolak ukur keberhasilan jawaban.
* **V (Verification):** Langkah-langkah pengujian kebenaran dan validasi hasil.


### Prompt 1 — Evaluasi Statistik Katalog Gempabumi
#### Role: Posisikan diri Anda sebagai peneliti seismologi yang bertugas membimbing mahasiswa S1 dalam mengeksplorasi pola dan karakteristik dari katalog gempabumi regional.

#### Task: Lakukan analisis statistik komprehensif terhadap basis data gempa untuk mengkaji sebaran spasial episenter, variasi kedalaman hiposenter, distribusi magnitudo, serta periodisitas waktu kejadian di daerah penelitian

#### Context: Katalog gempabumi menyimpan rekaman waktu kejadian, lokasi koordinat, kedalaman, dan magnitudo. Analisis ini dijalankan dalam mata kuliah Seismologi untuk memahami karakteristik seismisitas wilayah serta keterkaitannya dengan struktur tektonik aktif maupun zona subduksi.

#### Input: Gunakan data katalog gempa yang mencakup parameter waktu, koordinat lintang-bujur, kedalaman, dan skala magnitudo. Seluruh proses analisis wajib bersumber penuh dari data ini.

#### Constraints:
 ### Dilarang mengubah parameter asli dalam katalog.

#### Jangan mencampur skala magnitudo yang berbeda tanpa konversi atau penjelasan kontekstual.

#### Jelaskan penanganan data kosong atau data di luar batas area kajian.

#### Dilarang membuat prediksi spesifik terkait bencana gempa masa depan atau analisis tingkat risiko/kerentanan.

#### Hubungan antara pola seismisitas dengan tatanan tektonik lokal hanya boleh disampaikan sebagai hipotesis ilmiah yang didasari literatur/bukti data.

#### Output: Laporan analisis harus menyajikan:

#### Ringkasan deskriptif katalog (rentang waktu, batas koordinat wilayah, dan jumlah total kejadian gempa).

#### Analisis parameter statistik dasar untuk nilai magnitudo dan kedalaman.

#### Pola sebaran spasial episenter beserta penampang kedalaman (cross-section).

#### Grafik tren kronologis frekuensi kemunculan gempa.

#### Pembahasan naratif mengenai interpretasi pola kegempaan.

#### Kesimpulan singkat dan padat.

#### Quality Criteria: Seluruh nilai statistik harus sinkron dengan data katalog, istilah seismologi digunakan secara presisi, interpretasi tektonik proporsional, dan format tulisan sesuai standar tugas akademik.

#### Verification: Validasi ulang total kejadian (event), pastikan rentang magnitudo dan kedalaman terdata dengan benar, serta pastikan setiap poin kesimpulan terhubung langsung dengan bukti statistik katalog.

### Prompt 2 — Simulasi Model Termal Litosfer Samudera
#### Role: Bertindaklah sebagai pakar geodinamika dan pemodelan komputasi yang mendampingi mahasiswa S1 Geofisika dalam mempelajari proses pendinginan konduktif litosfer samudera.
#### Task: Bedah formulasi matematika dari pemodelan pendinginan konduktif 1D, kemudian bandingkan hasil simulasi numerik terhadap solusi analitik eksaknya pada berbagai variasi umur lempeng samudera.
#### Context: Seiring bertambahnya usia, litosfer samudera mengalami pendinginan, kontraksi termal, dan penebalan akibat pelepasan fluks panas ke permukaan. Materi ini dipelajari dalam mata kuliah Geodinamika dan Geofisika Komputasi untuk menghubungkan umur lempeng dengan profil temperatur terhadap kedalaman.
#### Input: Gunakan parameter konstanta yang telah ditentukan (seperti difusivitas termal, suhu permukaan, suhu mantel/astenosfer, batas kedalaman maksimum, dan beberapa variasi umur lempeng). Dilarang memasukkan variabel luar tanpa penjelasan.
#### Constraints:Jangan mengubah nilai parameter acuan tanpa catatan transparan. Uraikan asumsi penyederhanaan model (misal: mengabaikan generasi panas internal radioaktif dan efek konveksi).Pastikan konsistensi satuan dalam semua rumus, terutama konversi umur lempeng dari jutaan tahun (Ma) ke detik. Jelaskan syarat kestabilan komputasi numerik (seperti kriteria Courant-Friedrichs-Lewy / CFL).Jangan menyamakan begitu saja model teoretis sederhana ini dengan kondisi riil interior bumi yang jauh lebih kompleks.
#### Output: Luaran laporan harus menyajikan:Penjelasan persamaan diferensial pengatur beserta asumsi fisiknya dengan bahasa yang jelas.Penentuan kondisi awal (initial conditions) dan kondisi batas (boundary conditions).Grafik perbandingan profil temperatur terhadap kedalaman ($T \text{ vs } Z$) untuk berbagai variasi umur lempeng.Evaluasi komparatif antara metode numerik vs analitik beserta angka deviasi/galatnya.Penentuan kedalaman isoterm acuan (misal $1300^\circ\text{C}$) dan hubungannya dengan umur lempeng.Narasi interpretasi implikasi geodinamikannya.Evaluasi batasan model dan simpulan ringkas.
#### Quality Criteria: Dasar hukum fisika termal dijelaskan secara tepat, perbedaan nilai numerik dan analitik dihitung secara kuantitatif, satuan konsisten, dan interpretasi tetap berada dalam koridor asumsi model.
#### Verification: Periksa apakah selisih/galat numerik berkurang saat ukuran grid kedalaman diperhalus ($\Delta z$ lebih kecil), pastikan kondisi batas suhu permukaan dan dasar terpenuhi, serta pastikan kedalaman isoterm makin menunjam seiring bertambahnya umur lempeng.    

### Prompt 3 — Analisis Geostatistika & Pemetaan Spasial
#### Role: Ambil peran sebagai ahli geostatistika yang membimbing mahasiswa sarjana Geofisika dalam menganalisis dependensi spasial dan melakukan estimasi variabel melalui interpolasi.
#### Task: Analisis struktur korelasi spasial antar titik sampel menggunakan variogram eksperimental, jelaskan penerapannya dalam metode interpolasi Kriging, serta evaluasi tingkat ketidakpastian peta hasilnya.
#### Context: Data pengukuran geofisika di lapangan umumnya bersifat titik-titik diskrit, padahal penyajikan peta kontur membutuhkan estimasi yang kontinu. Topik ini diajarkan pada mata kuliah Geostatistika & Analisis Spasial untuk memahami tingkat kemiripan spasial sebelum membuat peta estimasi.
#### Input: Gunakan dataset yang memuat koordinat lokasi ($X, Y$) serta satu atribut fisika batuan (seperti anomali gayaberat, tebal sedimen, atau elevasi muka air tanah). Perhatikan sistem referensi koordinat serta satuannya.
#### Constraints:Pertahankan keaslian nilai pengukuran tanpa perubahan acak.Pastikan koordinat telah dikonversi ke sistem proyeksi bersatuan meter sebelum menghitung jarak spasial.Jangan menghapus nilai ekstrim (outlier) secara diam-diam dan jelaskan dampak keberadaannya terhadap model variogram.Berikan alasan ilmiah saat menentukan jarak lag (lag size) dan pemilihan model variogram teoretis.Hindari mengklaim akurasi tinggi pada wilayah ekstrapolasi (titik di luar cakupan data sampel).
#### Output: Laporan akhir wajib mencakup:Deskripsi karakteristik dataset dan statistik deskriptif awal.Bentuk variogram eksperimental beserta cara membaca dan menginterpretasikannya.Makna fisis dari variabel nugget effect, sill, dan range.Penentuan model teoretis variogram (misal: Spherical, Exponential, Gaussian) disertai alasannya.Peta hasil interpolasi Kriging beserta peta variansi ketidakpastian spasialnya.Perbandingan ringkas antara Kriging dengan metode deterministik (seperti IDW).Simpulan menyeluruh.
#### Quality Criteria: Penyampaian konsep geostatistik sistematis dan komunikatif, seluruh parameter mencantumkan satuan yang jelas, pembahasan keterbatasan data diuraikan secara terbuka, serta hasil siap dipakai dalam tugas.
#### Verification: Pastikan nilai range realistis terhadap dimensi area penelitian, terapkan validasi silang (cross-validation/leave-one-out), dan pastikan kesimpulan analisis selaras dengan hasil uji validasi tersebut.

### Prompt 4 — Pemodelan & Interpretasi Seismik Refraksi
#### Role: Bertindaklah sebagai konsultan senior geofisika eksplorasi dangkal yang sedang mendampingi mahasiswa S1 Geofisika dalam menganalisis data waktu tempuh gelombang seismik.
#### Task: Lakukan evaluasi mendalam pada data first arrival (waktu tiba pertama) untuk menentukan profil kecepatan gelombang di tiap perlapisan serta menghitung kedalaman bidang batas litologi di bawah permukaan.
#### Context: Metode seismik refraksi dangkal sering dimanfaatkan untuk memetakan struktur geologi dekat permukaan, seperti mengidentifikasi batas antara lapisan tanah lapuk dan batuan dasar masif. Tugas ini merupakan bagian dari praktikum Metode Seismik Refraksi untuk mengasah keahlian mahasiswa dalam mengolah dan menginterpretasi grafik waktu-jarak ($T-X$).
#### Input: Gunakan lembar data yang berisi variabel jarak geofon dari sumber getaran (offset) serta catatan waktu kedatangan awal gelombang. Perhatikan secara cermat konsistensi satuan waktu dan jarak yang digunakan.
#### Constraints:Dilarang keras merubah angka-angka pada dataset secara sepihak.Uraikan langkah-langkah pemisahan segmen titik kurva menjadi gelombang langsung (direct wave) dan gelombang bias (refracted wave) secara logis.Jelaskan asumsi dasar yang dipakai (misalnya model lapisan datar dengan kecepatan bertambah terhadap kedalaman).Apabila ditemukan data yang menyimpang dari tren (outlier), berikan analisis kritis tanpa langsung menghapusnya.Jangan langsung memvonis jenis batuan secara absolut hanya berpatokan pada besaran nilai kecepatan gelombang.
#### Output: Sajikan laporan terstruktur yang memuat:Penjelasan ringkas mengenai karakteristik dataset dan satuan ukur.Prosedur plotting diagram $T-X$ serta teknik pemisahan segmen gelombang.Nilai kecepatan tiap lapisan beserta rumus penurunannya.Perhitungan estimasi kedalaman/ketebalan lapisan pertama beserta formulasinya.Deskripsi naratif interpretasi kondisi geologi bawah permukaan secara objektif.Evaluasi keterbatasan metode dan ringkasan simpulan.
#### Quality Criteria: Perhitungan numerik harus tepat dan konsisten secara satuan, alur penjelasan runtut dan mudah dipahami mahasiswa tingkat sarjana, interpretasi berbasis fakta data, serta format dokumen siap dilampirkan dalam laporan praktikum.
#### Verification: Pastikan nilai kecepatan lapisan bawah lebih besar dari lapisan di atasnya ($v_2 > v_1$), periksa konsistensi titik potong (crossover distance), dan pastikan kisaran nilai kecepatan berada dalam batas logis material geologi di alam.

### Prompt 5 — Klasifikasi Litologi Sumur Menggunakan Pembelajaran Mesin
#### Role: Posisikan diri Anda sebagai spesialis Data Science & AI Geosains yang mendampingi mahasiswa sarjana Geofisika dalam mengaplikasikan algoritma Machine Learning (ML) pada data rekaman log sumur.

#### Task: Rancang, bangun, dan evaluasi performa model pembelajaran mesin untuk memprediksi/mengklasifikasi jenis litologi berdasarkan data log sumur, serta bandingkan hasilnya terhadap model acuan (baseline) sederhana.

#### Context: Data log pemboran (seperti Gamma Ray, Density, Neutron Porosity, dan Sonic) memuat pola khas yang mencerminkan karakteristik batuan bawah permukaan. Analisis ini dipelajari pada mata kuliah AI untuk Geofisika guna memahami potensi dan kendala penerapan algoritma cerdas dalam interpretasi geologi.

#### Input: Gunakan dataset yang mencakup ID sumur, kedalaman, rekaman kurva log geofisika, dan label litologi acuan. Cermati satuan dari setiap kurva log serta keabsahan label litologinya.

#### Constraints:Dilarang memodifikasi nilai data mentah tanpa metodologi preprocessing yang sah.Pembagian data latih (train) dan data uji (test) wajib dilakukan berbasis sumur (well-based split), bukan pengacakan baris data (random row split), disertai alasan ilmiahnya.Dilarang keras membiarkan informasi dari data uji bocor saat proses hyperparameter tuning.Atribut kedalaman (depth) dan nama/ID sumur tidak boleh dijadikan fitur masukan (input features) dalam pemodelan.Jelaskan penanganan nilai kosong (null/missing values) atau data anomali secara fisika batuan. Jangan mengklaim bahwa model yang dibangun dapat langsung digeneralisasi ke seluruh lapangan minyak dan gas lain secara universal.

#### Output: Laporan akhir wajib berisi:Ringkasan deskripsi dataset dan analisis ketidakseimbangan proporsi kelas litologi (class imbalance).Arsitektur model baseline (misal: Naive Bayes / Decision Tree sederhana) dan model ML utama (misal: Random Forest / XGBoost).Evaluasi performa yang mendalam per kelas litologi (menggunakan Precision, Recall, F1-score), tidak hanya akurasi keseluruhan (overall accuracy). Analisis matriks kebingungan (confusion matrix / error analysis) untuk mengidentifikasi litologi yang sering tertukar beserta alasan geologisnya. Diskusi mengenai dampak ketidakseimbangan jumlah data tiap kelas terhadap prediksi model. Interpretasi naratif dan kesimpulan padat.

#### Quality Criteria: Evaluasi terbebas dari kebocoran data (data leakage), performa dirinci transparan per kelas, penjelasan mudah dipahami mahasiswa tingkat sarjana, serta hasil dapat langsung disematkan ke laporan akhir.

#### Verification: Pastikan tidak ada data dari sumur yang sama hadir di training set sekaligus test set, verifikasi bahwa model ML utama mengungguli model baseline, dan pastikan performa model anjlok secara signifikan apabila label litologi diacak (label shuffling test).
